```python
import os
from dotenv import load_dotenv

load_dotenv()

api_key = os.getenv("MY_API_KEY")
Server = os.getenv("DB_SERVER")
username = os.getenv("DB_USERNAME")
password = os.getenv("DB_PASSWORD")
```

```python
cities = [
    "Cairo",
    "Giza",
    "Alexandria",
    "Port Said",
    "Suez",
    "Ismailia",
    "Mansoura",
    "Tanta",
    "Luxor",
    "Aswan"
]
```

```python
import requests
url = "https://api.openweathermap.org/data/2.5/weather"
```

```python
info=[]
for city in cities:
    params = {
    "appid": api_key,
    "q": f'{city}'
    }

    response = requests.get(url, params=params)
    data = response.json()
    data
    info.append(
        {
    'country':'Egypt',
    'city':f'{city}',
    'Longitude':data['coord']['lon'],
    'Latitude':data['coord']['lat'],
    'temprature':data['main']['temp'],
    'humidity':data['main']['humidity'],
    'pressure':data['main']['pressure']
    })
```

```python
import pandas as pd
df = pd.DataFrame(info)
df
```

Transformaiton

```python
df['temp_in_cel']=df['temprature']-273
```

```python
df
```

```python
  * Greater Cairo
  * North Coast
  * Lower Egypt
  * Upper Egypt
  * Canal
  * Sinai
  * Red Sea
```

```python
def func1(city):
    if city=='Cairo':
        return 'Greater Cairo'
    elif city=='Giza':
        return 'Greater Cairo'
    elif city=='Alexandria':
        return 'Lower Egypt'
    elif city== 'Port Said':
        return 'Canal'
    elif city== 'Suez':
        return 'Canal'
    elif city== 'Ismailia':
            return 'Sinai'
    elif city== 'Mansoura':
                return 'Sinai'
    elif city== 'Tanta':
                return 'Red Sea'
    elif city== 'Luxor':
                return 'Upper Egypt'
    elif city== 'Aswan':
                    return 'Upper Egypt'
```

```python
df['region']=df['city'].apply(func1)
```

```python
df
```

```python
df['TemperatureCategory']=df['temp_in_cel'].apply(lambda x : 'hot' if x >=30 else 'cold' )
```

```python
df
```

```python
df['humidityCategory']=df['humidity'].apply(lambda x : 'high' if x >=30 else 'low' )
```

```python
import datetime
df['IngestionTime'] = datetime.datetime.now()
```
